# Import Library

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

# Import Dataset

````python
dataset = pd.read_csv("Position_Salaries.csv")
X = dataset.iloc[:,1:-1].values
y = dataset.iloc[:,-1].values.reshape(10,1)
````

# Feature Scaling

````python
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
sc_y = StandardScaler()
X = sc_X.fit_transform(X)
y = sc_y.fit_transform(y)
````

# Training the SVR model on the whole dataset


````python
from sklearn.svm import SVR
Svr_reg = SVR(kernel = 'rbf')
Svr_reg.fit(X,y)
````

# Training the Support Vector Regression (SVR) Model

```python
from sklearn.svm import SVR

svr_reg = SVR(kernel='rbf')
svr_reg.fit(X, y)
```

The actual learning starts when the following line is executed:

```python
svr_reg.fit(X, y)
```

At this stage, the SVR model learns the relationship between the input feature (**Level**) and the target variable (**Salary**).

---

# Step 1: Receive the Training Dataset

The model first receives all the training data.

| Level (X) | Salary (Y) |
|-----------|-----------:|
|1|45,000|
|2|50,000|
|3|60,000|
|4|80,000|
|5|110,000|
|6|150,000|
|7|200,000|
|8|300,000|
|9|500,000|
|10|1,000,000|

If feature scaling has been performed before training, the model receives the **scaled values** instead of the original values.

At this stage, the model has **not learned anything**.

---

# Step 2: Initialize the SVR Model

The model creates an empty Support Vector Regression model.

It stores the chosen hyperparameters such as:

- Kernel (RBF)
- C (Penalty parameter)
- ε (Epsilon)
- Gamma

Initially, the model has:

- No regression function
- No support vectors
- No coefficients
- No bias

The model is simply ready to start learning.

---

# Step 3: Create an Initial Regression Function

The optimizer creates an initial regression function.

Initially, this regression function is only a starting point and does not fit the data properly.

Using this regression function, the model predicts a salary for every training sample.

These initial predictions are usually inaccurate because the model has not yet been optimized.

---

# Step 4: Create the ε-Insensitive Tube

Suppose

```
ε = 20,000
```

For every predicted salary, SVR creates an acceptable region called the **ε-insensitive tube**.

If the prediction for a training sample is

```
150,000
```

then

Upper Boundary

```
150,000 + 20,000

=

170,000
```

Lower Boundary

```
150,000 - 20,000

=

130,000
```

Any actual salary between

```
130,000

↓

170,000
```

is considered acceptable.

The upper and lower boundaries are created around **every prediction**, so they change whenever the prediction changes.

---

# Step 5: Compare Every Training Sample

The model now compares every actual salary with its predicted salary.

For every training sample,

```
Error = |Actual Salary - Predicted Salary|
```

Now SVR checks the error.

### Case 1

If

```
Error ≤ ε
```

then the point lies inside the ε-tube.

Penalty

```
0
```

The model completely ignores this point because the prediction is already close enough.

---

### Case 2

If

```
Error > ε
```

then the point lies outside the ε-tube.

Only the part of the error outside the ε-tube is penalized.

Penalty

```
Penalty = Error − ε
```

For example,

if

```
Error = 30,000

ε = 20,000
```

then

```
Penalty

=

30,000 − 20,000

=

10,000
```

Notice that the entire error is **not** penalized.

Only the amount outside the ε-tube contributes to the loss.

---

# Step 6: Calculate the Total Penalty

The above process is repeated for every training sample.

For each sample,

```
If Error ≤ ε

Penalty = 0

Else

Penalty = Error − ε
```

The penalty of every training sample is added together.

```
Total Penalty

=

Penalty₁

+

Penalty₂

+

Penalty₃

+

...

+

Penaltyₙ
```

This total penalty tells the optimizer how good or bad the current regression function is.

A smaller total penalty means a better regression function.

---

# Step 7: Optimize the Regression Function

Now the optimizer starts improving the regression function.

Its objective is to:

- Reduce the total penalty.
- Keep as many training samples as possible inside the ε-tube.
- Keep the regression function as smooth as possible.
- Avoid creating an unnecessarily complex model.

Internally, Scikit-Learn uses **Quadratic Programming (LIBSVM solver)** to perform this optimization.

The optimizer repeatedly updates the model parameters until it finds the regression function with the minimum objective value.

This entire optimization process happens automatically inside the `fit()` function.

---

# Step 8: Identify the Support Vectors

After optimization, most training samples lie inside the ε-tube.

These samples have zero penalty.

Therefore, they no longer influence the regression function.

Only the samples that

- lie outside the ε-tube, or
- lie exactly on the ε-tube boundary

remain important.

These important training samples are called **Support Vectors**.

Support vectors determine the final position and shape of the regression function.

---

# Step 9: Store the Learned Model

After training is complete, the model stores:

- Support Vectors
- Dual coefficients (weights)
- Bias (Intercept)
- Kernel information
- Hyperparameters
- Scaling information (if scaling was performed)

The model does **not** need every training sample for future predictions.

Most predictions depend mainly on the support vectors.

---

# Step 10: Training Completes

When the optimizer finds the best possible regression function,

```python
svr_reg.fit(X, y)
```

finishes executing.

At this point, the model has learned the relationship between **Level** and **Salary**.

The trained model is now ready to predict the salary for any new employee.

Example:

```python
svr_reg.predict([[6.5]])
```

The model scales the input (if scaling was used), compares it with the learned support vectors using the RBF kernel, computes the predicted salary, converts it back to the original salary scale (inverse scaling), and returns the final prediction.

---

# Complete Flow of Training

```
Training Dataset

        │
        ▼

Initialize Empty SVR Model

        │
        ▼

Create Initial Regression Function

        │
        ▼

Predict Every Training Sample

        │
        ▼

Create ε-Tube Around Predictions

        │
        ▼

Calculate Prediction Error

        │
        ▼

If Error ≤ ε

Penalty = 0

Else

Penalty = Error − ε

        │
        ▼

Calculate Total Penalty

        │
        ▼

Optimize Regression Function
(Quadratic Programming)

        │
        ▼

Identify Support Vectors

        │
        ▼

Store Learned Parameters

        │
        ▼

Training Complete
```

# Key Points to Remember

- SVR does **not** try to fit every training sample exactly.
- Errors inside the ε-tube are completely ignored.
- Only the part of the error outside the ε-tube is penalized.
- Only support vectors determine the final regression function.
- The optimizer automatically learns the best regression function by minimizing the total penalty while keeping the model smooth.
- 

````python
Svr_reg.predict(sc_X.transform([[6.5]]))
````

````python
sc_y.inverse_transform(Svr_reg.predict(sc_X.transform([[6.5]])).reshape(-1,1))
````

````python
plt.scatter(sc_X.inverse_transform(X), sc_y.inverse_transform(y), color='red')

plt.plot(
    sc_X.inverse_transform(X),
    sc_y.inverse_transform(Svr_reg.predict(X).reshape(-1, 1)),
    color='blue'
)

plt.title('Truth or Bluff (SVR)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
````

# Visualising the SVR results (for higher resolution and smoother curve)

````python
# Create a high-resolution grid
X_grid = np.arange(
    sc_X.inverse_transform(X).min(),
    sc_X.inverse_transform(X).max(),
    0.1
).reshape(-1, 1)

# Plot original observations
plt.scatter(
    sc_X.inverse_transform(X),
    sc_y.inverse_transform(y),
    color='red'
)

# Plot the SVR regression curve
plt.plot(
    X_grid,
    sc_y.inverse_transform(
        Svr_reg.predict(sc_X.transform(X_grid)).reshape(-1, 1)
    ),
    color='blue'
)

plt.title('Truth or Bluff (SVR)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
````
