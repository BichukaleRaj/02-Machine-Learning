# Meaning of Multiple Linear Regression

**Multiple Linear Regression (MLR)** is a supervised machine learning algorithm used to predict the value of a **dependent variable** based on **two or more independent variables**.

Unlike Simple Linear Regression, which uses only one independent variable, Multiple Linear Regression considers the combined effect of multiple input features to make more accurate predictions.

# Multiple Linear Regression

## Expression

$$
\hat{y}=b_0+b_1x_1+b_2x_2+b_3x_3+\cdots+b_nx_n
$$

## Where,

- $\hat{y}$ → Predicted (Dependent Variable)
- $b_0$ → Intercept (Value of $\hat{y}$ when all independent variables are 0)
- $b_1,b_2,b_3,\ldots,b_n$ → Regression Coefficients (Change in $\hat{y}$ for a one-unit increase in the corresponding independent variable, keeping all other variables constant)
- $x_1,x_2,x_3,\ldots,x_n$ → Independent Variables (Input Features)
- $n$ → Total number of independent variables (features)

# 01.Understanding Dummy Variables

Suppose our dataset contains a categorical column called **State**.

| State |
|--------|
| New York |
| California |
| New York |
| California |

Since a Machine Learning model **cannot understand text values**, we convert them into numbers using **Dummy Variables**.

---

# Step 1: Create Dummy Variables

| State | New York | California |
|--------|----------|------------|
| New York | 1 | 0 |
| California | 0 | 1 |
| New York | 1 | 0 |
| California | 0 | 1 |

Here,

- **1** → The company belongs to that state.
- **0** → The company does not belong to that state.

---

# Step 2: Remove One Dummy Variable

Instead of using both dummy variables,

| New York | California |
|----------|------------|
|1|0|
|0|1|
|1|0|
|0|1|

we keep only **one dummy variable**.

For example,

| New York |
|----------|
|1|
|0|
|1|
|0|

The **California** column is removed.

---

# Why don't we lose information?

Even after removing the California column, we still know the state.

| New York | Meaning |
|----------|---------|
|1|Company is in **New York**|
|0|Company is in **California**|

Therefore,

- **1** automatically means **New York**.
- **0** automatically means **California**.

Hence, **no information is lost**.

---

# Dummy Variable Works Like a Light Switch

Think of the dummy variable as an **ON/OFF switch**.

### Switch ON

```
New York = 1
```

➡ Company is in **New York**

---

### Switch OFF

```
New York = 0
```

➡ Company is in **California**

Thus, a dummy variable simply tells the model whether a category is present or not.

---

# Regression Equation

Suppose the regression equation is

$$
\text{Profit}
=
b_0
+
b_1(\text{R\&D})
+
b_2(\text{Admin})
+
b_3(\text{Marketing})
+
b_4(\text{New York})
$$

> **Note:** Only **one dummy variable (New York)** is included.

---

# Case 1: Company is in California

For California,

$$
\text{New York}=0
$$

Substituting into the equation,

$$
\text{Profit}
=
b_0
+
b_1(\text{R\&D})
+
b_2(\text{Admin})
+
b_3(\text{Marketing})
+
b_4(0)
$$

Since

$$
b_4 \times 0 = 0
$$

the equation becomes

$$
\text{Profit}
=
b_0
+
b_1(\text{R\&D})
+
b_2(\text{Admin})
+
b_3(\text{Marketing})
$$

This becomes the **default equation** for **California**.

---

# Case 2: Company is in New York

For New York,

$$
\text{New York}=1
$$

Substituting into the equation,

$$
\text{Profit}
=
b_0
+
b_1(\text{R\&D})
+
b_2(\text{Admin})
+
b_3(\text{Marketing})
+
b_4(1)
$$

Since

$$
b_4 \times 1 = b_4
$$

the equation becomes

$$
\text{Profit}
=
b_0
+
b_1(\text{R\&D})
+
b_2(\text{Admin})
+
b_3(\text{Marketing})
+
b_4
$$

Compared to California, an **extra coefficient ($b_4$)** is added.

This adjusts the prediction for companies located in **New York**.

---

# Why is this not biased?

At first, it may seem that only **New York** has a coefficient (**$b_4$**), while **California** has none.

However, this is **not biased**.

The regression model automatically treats **California** as the **default category**.

- When **New York = 0**, the equation represents **California**.
- When **New York = 1**, the model adds **$b_4$** to adjust the equation for **New York**.

Therefore,

- **California** → Default Equation
- **New York** → Default Equation + **$b_4$**

Both states are represented correctly.

---

# Final Idea

A dummy variable behaves like an **ON/OFF switch**.

| Dummy Variable | Meaning |
|---------------|---------|
| **0** | Use the **default category (California)** |
| **1** | Add the coefficient **$b_4$** to switch to **New York** |

Therefore, **only one dummy variable is enough to represent two categories without losing any information.**

# 02.Dummy Variable Trap

The **Dummy Variable Trap** occurs when **all dummy variables** of a categorical feature are included in a Linear Regression model.

Since one dummy variable can be **calculated from the others**, the model receives **duplicate information**, causing **Multicollinearity**.

---

# Example

Suppose the categorical feature **State** has two categories.

| State | New York (D₁) | California (D₂) |
|--------|---------------|-----------------|
| New York | 1 | 0 |
| California | 0 | 1 |
| New York | 1 | 0 |
| California | 0 | 1 |

Notice that

$$
D_2 = 1 - D_1
$$

This means that if we already know **D₁**, we can always calculate **D₂**.

Therefore, both dummy variables contain the **same information**.

---

# Regression Equation

Suppose the regression equation is

$$
y
=
b_0
+
b_1x_1
+
b_2x_2
+
b_3x_3
+
b_4D_1
+
b_5D_2
$$

Since

$$
D_2 = 1 - D_1
$$

Substituting,

$$
y
=
b_0
+
b_1x_1
+
b_2x_2
+
b_3x_3
+
b_4D_1
+
b_5(1-D_1)
$$

Expanding,

$$
y
=
b_0
+
b_1x_1
+
b_2x_2
+
b_3x_3
+
b_4D_1
+
b_5
-
b_5D_1
$$

Rearranging,

$$
y
=
(b_0+b_5)
+
b_1x_1
+
b_2x_2
+
b_3x_3
+
(b_4-b_5)D_1
$$

Now the model cannot uniquely determine the values of **$b_0$**, **$b_4$**, and **$b_5$** because the dummy variables contain duplicate information.

This problem is called the **Dummy Variable Trap**.

---

# What is Multicollinearity?

**Multicollinearity** occurs when one independent variable can be **predicted using another independent variable**.

Here,

$$
D_2 = 1 - D_1
$$

Therefore,

- Knowing **New York (D₁)** automatically tells us **California (D₂)**.
- Both variables provide the same information.
- The regression model cannot distinguish their individual effects.

---

# How to Avoid the Dummy Variable Trap?

Simply remove **one dummy variable**.

Instead of

| New York | California |
|----------|------------|
|1|0|
|0|1|
|1|0|

use only

| New York |
|----------|
|1|
|0|
|1|

Now,

- **New York = 1** → Company is in **New York**
- **New York = 0** → Company is in **California**

No information is lost.

---

# General Rule

If a categorical variable has **n categories**, include only **n − 1 dummy variables**.

| Number of Categories | Dummy Variables to Include |
|----------------------|----------------------------|
|2|1|
|3|2|
|4|3|
|5|4|
|10|9|
|100|99|

---

# If Multiple Categorical Variables Exist

Apply the **n − 1 rule separately** to each categorical feature.

### Example

| Categorical Feature | Categories | Dummy Variables to Include |
|---------------------|-----------:|---------------------------:|
| State | 3 | 2 |
| Industry | 5 | 4 |

---

# Final Idea

The **Dummy Variable Trap** occurs because one dummy variable can be **completely calculated from another**, creating **perfect multicollinearity**.

To avoid this problem, always include **only n − 1 dummy variables** for a categorical variable having **n categories**.

# 03.P values and statistical significance

# Statistical Significance and P-Value

## What is Hypothesis Testing?

Hypothesis Testing is a statistical method used to decide whether a statement (hypothesis) is likely to be true or false based on the observed data.

There are two hypotheses:

- **Null Hypothesis ($H_0$):** Assumes that there is **no significant effect or difference**. It is considered **true initially**.
- **Alternative Hypothesis ($H_1$ or $H_a$):** Assumes that there **is a significant effect or difference**.

---

## Coin Toss Example

Suppose we assume that a coin is **fair**.

$$
H_0:\ \text{The coin is fair}
$$

$$
H_1:\ \text{The coin is not fair}
$$

Initially, we always assume that **$H_0$ is true**.

---

## What is the P-Value?

The **P-value** is the probability of obtaining the observed result **assuming that the Null Hypothesis ($H_0$) is true**.

As unusual events occur, the P-value becomes smaller.

Example:

| Result | P-Value |
|---------|---------|
| 1 Tail | 50% |
| 2 Tails | 25% |
| 3 Tails | 12.5% |
| 4 Tails | 6.25% |
| 5 Tails | 3.125% |

As the number of consecutive tails increases, the event becomes less likely if the coin is actually fair.

---

## Statistical Significance

A significance level (**$\alpha$**) is chosen before performing the test.

Usually,

$$
\alpha = 0.05 \; (5\%)
$$

Now compare the **P-value** with **$\alpha$**.

### Case 1

If

$$
P\text{-value} > 0.05
$$

The result is **not statistically significant**.

➡ Do **not reject** the Null Hypothesis ($H_0$).

---

### Case 2

If

$$
P\text{-value} \le 0.05
$$

The result is **statistically significant**.

➡ Reject the Null Hypothesis ($H_0$).

➡ Accept the Alternative Hypothesis ($H_1$).

---

# Final Idea

- Start by assuming **$H_0$ is true**.
- Perform the experiment.
- Calculate the **P-value**.
- Compare the P-value with **0.05**.
- If the P-value is **small (≤ 0.05)**, reject **$H_0$** because the observed result is too unlikely to occur by chance.
- If the P-value is **greater than 0.05**, keep **$H_0$** because there is not enough evidence to reject it.

# Methods of Building a Multiple Linear Regression Model

When a dataset contains multiple independent variables (features), **not all variables contribute equally** to predicting the dependent variable. Therefore, feature selection methods are used to identify the most significant variables and build an efficient regression model.

There are **five commonly used methods** for building a Multiple Linear Regression model.

---

# 1. All-in Method

### Meaning

In the **All-in Method**, **all independent variables** are included in the regression model.

No feature is removed, and every available predictor is used for training the model.

### When is it used?

- When all variables are believed to be important.
- When feature selection is not required.
- When the objective is to study the effect of every variable.

---

# 2. Backward Elimination

### Meaning

Backward Elimination starts with **all independent variables** in the model.

The variable with the **highest P-value** (least significant variable) is removed first.

The process is repeated until all remaining variables have **P-values less than the chosen significance level**.

### Purpose

To remove insignificant variables and keep only the important predictors.

---

# 3. Forward Selection

### Meaning

Forward Selection starts with **no independent variables**.

The most significant variable is added first.

The remaining variables are added **one by one** only if they improve the regression model.

The process stops when adding another variable does not improve the model.

### Purpose

To build the regression model gradually using only significant variables.

---

# 4. Bidirectional Elimination (Stepwise Regression)

### Meaning

Bidirectional Elimination combines **Forward Selection** and **Backward Elimination**.

Variables are added one at a time, and after each addition, the model checks whether any previously selected variable has become insignificant.

If a variable becomes insignificant, it is removed.

This process continues until the best combination of variables is obtained.

### Purpose

To continuously add useful variables and remove unnecessary variables to obtain an optimized regression model.

---

# 5. Score Comparison

### Meaning

In the Score Comparison method, multiple regression models are built using different combinations of variables.

The performance of each model is evaluated using performance metrics such as:

- R² Score
- Adjusted R²
- AIC
- BIC

The model with the best performance score is selected.

### Purpose

To choose the regression model that provides the highest prediction accuracy.

---

# What is Stepwise Regression?

**Stepwise Regression** is a feature selection technique in which variables are selected **step by step**.

It includes:

- Backward Elimination
- Forward Selection
- Bidirectional Elimination

These methods gradually **add**, **remove**, or **both add and remove** variables to obtain the best regression model.

---

# Summary

| Method | Description |
|---------|-------------|
| **All-in** | Includes all independent variables in the model. |
| **Backward Elimination** | Starts with all variables and removes insignificant variables one by one. |
| **Forward Selection** | Starts with no variables and adds significant variables one by one. |
| **Bidirectional Elimination** | Adds significant variables and removes insignificant variables until the best model is obtained. |
| **Score Comparison** | Compares multiple regression models and selects the one with the best performance score. |
