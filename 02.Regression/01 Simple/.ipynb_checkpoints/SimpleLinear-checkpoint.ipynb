# Meaning of Simple Linear Regression

**Simple Linear Regression (SLR)** is a supervised machine learning algorithm used to predict the value of a **dependent variable** based on **one independent variable**.

It finds the **best-fit straight line** that represents the relationship between the independent variable and the dependent variable, allowing the model to make predictions for new input values.

# Simple Linear Regression

## Expression

$$
\hat{y}=b_0+b_1x
$$

### Where,

- $\hat{y}$ → Predicted (Dependent Variable)
- $b_0$ → Intercept (Value of $y$ when $x=0$)
- $b_1$ → Slope (Change in $y$ for every 1-unit increase in $x$)
- $x$ → Independent Variable (Input Feature)

```

# Training the Simple Linear Regression Model

## Step 1: Split the Dataset

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=0
)
```

### What happens?

- The dataset contains **30 rows**.
- `test_size = 0.2` means **20% of the data** is reserved for testing.

```
30 × 0.2 = 6 rows (Testing Data)

30 - 6 = 24 rows (Training Data)
```

After splitting:

- **X_train** → Experience values used for training.
- **y_train** → Salary values used for training.
- **X_test** → Experience values reserved for testing.
- **y_test** → Actual salaries reserved for comparison later.

> **Important:**  
> The model learns **only from the training data**.  
> The testing data is **never used during training**.

---

## Step 2: What does `random_state = 0` do?

Before splitting, the dataset is **shuffled randomly**.

Example:

Before shuffling

```
Row 1
Row 2
Row 3
...
Row 30
```

After shuffling

```
Row 18
Row 4
Row 27
Row 9
...
```

Then,

- First **24 rows** → Training Set
- Remaining **6 rows** → Testing Set

Setting

```python
random_state = 0
```

ensures that the dataset is shuffled **the same way every time** the notebook is executed.

---

# Step 3: Import the Linear Regression Class

```python
from sklearn.linear_model import LinearRegression
```

### What happens?

Python imports the **LinearRegression** class from the Scikit-Learn library.

At this stage:

- No model is created.
- No calculations are performed.

The class is simply loaded into memory.

---

# Step 4: Create the Model

```python
LR = LinearRegression()
```

### What happens?

A new Linear Regression model object is created.

At this moment, the model is completely empty.

```
LR

Slope (m)      = ?
Intercept (c)  = ?
Status         = Not Trained
```

The model is ready to learn but has not learned anything yet.

---

# Step 5: Train the Model

```python
LR.fit(X_train, y_train)
```

This is the most important step.

The model starts learning **only from**

- `X_train`
- `y_train`

The testing data

- `X_test`
- `y_test`

is completely ignored during training.

---

# Step 6: Read the Training Data

The model reads every row from the training dataset.

Example:

| Experience | Salary |
|------------|--------|
| 2.2 | 39891 |
| 5.1 | 66029 |
| 3.0 | 60150 |
| 8.7 | 109431 |
| ... | ... |

It continues until all **24 training rows** have been processed.

---

# Step 7: Calculate the Average Values

The model calculates

### Average Experience

$$
\bar{X}
$$

and

### Average Salary

$$
\bar{Y}
$$

using only the **training dataset**.

These averages are required for the next calculations.

---

# Step 8: Calculate the Slope (m)

The model uses the **Least Squares Formula**.

$$
m=\frac{\sum (X-\bar{X})(Y-\bar{Y})}{\sum (X-\bar{X})^2}
$$

### Internally, the model performs the following steps:

For every training row, it calculates

$$
m=\frac{\sum (X-\bar{X})(Y-\bar{Y})}{\sum (X-\bar{X})^2}
$$

Then,

- multiplies both values

$$
(X-\bar{X})(Y-\bar{Y})
$$

- adds all these values together.

This becomes the **Numerator**.

Next, it calculates

$$
(X-\bar{X})^2
$$

for every training row.

Then it adds all these squared values together.

This becomes the **Denominator**.

Finally,

```
Slope (m)

=

Numerator

÷

Denominator
```

This gives the value of **m**.

> **Important:**  
> The computer **does not guess** the value of **m**.
> It directly calculates it using the Least Squares Formula.

---

# Step 9: Calculate the Intercept (c)

After finding the slope, the model calculates

$$
c=\bar{Y}-m\bar{X}
$$

where

- X̄ = Average Experience
- Ȳ = Average Salary
- \(m\) = Slope

This gives the value of **c**.

Again, **no guessing is involved**.

---

# Step 10: Obtain the Best-Fit Line

Suppose the calculations produce

```
m = 9312.58

c = 26780.10
```

The learned equation becomes

$$
\boxed{\text{Salary}=9312.58\times\text{Experience}+26780.10}
$$

This equation is called the **Best-Fit Line**.

---

# Why is it called the Best-Fit Line?

There are infinitely many possible straight lines.

Example:

```
Salary = 7000X + 30000

Salary = 8000X + 25000

Salary = 9000X + 20000

Salary = 10000X + 15000

...
```

The Least Squares Formula guarantees that the line obtained has the **minimum Sum of Squared Errors (SSE)** among **all possible straight lines**.

> **Important:**  
> It is the **best possible straight line**, **not necessarily a perfect line**.

If the data itself does not follow a straight-line pattern, even the best straight line may still produce prediction errors.

---

# Step 11: Store the Learned Parameters

After training, the model stores the learned values inside the object.

```
LR

Slope (m)      = 9312.58

Intercept (c)  = 26780.10

Status         = Trained ✔
```

The training process is now complete.

The model has learned the relationship between

- Experience
- Salary

using only the training dataset.

---

## At this stage

- ✅ The model has been trained.
- ✅ The best-fit line has been obtained.
- ✅ The values of **m** and **c** have been stored inside the model.
- ✅ The testing data has **not been used yet**.
- ⏳ The next step will be to use this learned equation to predict salaries for the unseen testing data using `predict()`.

# Predicting the Test set results

````python
y_predict=LR.predict(X_test)
````

## Predicting the Test Set Results

```python
y_pred = LR.predict(X_test)
```

### What happens?

- During `LR.fit(X_train, y_train)`, the model learned the **best-fit line (equation)** using the training data.
- The values of **m (slope)** and **c (intercept)** are already stored inside the model.
- `LR.predict(X_test)` uses the **Experience values** from `X_test`.
- For each Experience value, the model substitutes it into the learned equation

$$
\text{Salary} = m \times \text{Experience} + c
$$

to calculate the **predicted Salary**.
- All predicted salaries are stored in an array called **`y_pred`**.

### Why do we use `X_test`?

- `X_test` contains **new (unseen) Experience values**.
- We use it to check whether the model can correctly predict salaries for data it has **never seen before**.

### Why don't we use `y_test`?

- `y_test` contains the **actual salaries**.
- The model predicts salaries using only `X_test`.
- After prediction, `y_pred` is compared with `y_test` to measure the model's accuracy.

# Visualizing the training sets result

````python
plt.scatter(X_train,y_train,color='red')
plt.plot(X_train,LR.predict(X_train),color='blue')
plt.title('Salary vs Experience(Training set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
````

```python
plt.scatter(X_train, y_train, color='red')

### What happens?

- `plt.scatter()` creates a **scatter plot**.
- It takes the values from `X_train` as the **X-coordinates** (Years of Experience).
- It takes the values from `y_train` as the **Y-coordinates** (Actual Salary).
- For every `(Experience, Salary)` pair, one **red dot** is plotted on the graph.

Example:

| Experience (`X_train`) | Salary (`y_train`) |
|------------------------|--------------------|
| 2.2 | 39891 |
| 5.1 | 66029 |
| 8.7 | 109431 |

These become three red points:

```

   •
        •

              •

-----------------------------> Experience
```

Thus, the scatter plot shows the **actual training data**.

```python
plt.plot(X_train, LR.predict(X_train), color='blue')
```

### What happens?

Python first executes

```python
LR.predict(X_train)
```

The model already knows the best-fit equation learned during training.

$$
\text{Salary}=m\times\text{Experience}+c
$$

Now, for every Experience value in `X_train`, the model substitutes it into this equation and predicts the corresponding Salary.

Example:

| Experience (`X_train`) | Predicted Salary |
|------------------------|------------------|
|2.2|47100|
|5.1|74200|
|8.7|107800|

Now `plt.plot()` receives

```python
plt.plot(
    X_train,
    Predicted_Salary,
    color='blue'
)
```

So the graph contains these points:

| X-coordinate | Y-coordinate |
|--------------|--------------|
|2.2|47100|
|5.1|74200|
|8.7|107800|

Instead of showing separate dots, `plt.plot()` **joins all these predicted points with a line**.

Since every predicted salary is obtained from the same linear equation

$$
\text{Salary}=m\times\text{Experience}+c
$$

all predicted points lie on a straight line.

Therefore, joining these points forms the **Best-Fit (Regression) Line**, shown in **blue**.

---

### Interpretation

- 🔴 **Red Dots** → Actual Salary values from the training dataset.
- 🔵 **Blue Line** → Predicted Salary values obtained using the learned equation.

By comparing the red dots with the blue line, we can visually see how well the learned regression line fits the training data.
