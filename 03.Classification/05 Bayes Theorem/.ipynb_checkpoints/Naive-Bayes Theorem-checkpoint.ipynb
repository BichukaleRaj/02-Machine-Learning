# Import Libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

# Import Dataset

````python
dataset = pd.read_csv("Social_Network_Ads.csv")
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:,-1].values
````

# Splitting the dataset into the Training and Test data

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
````

# Feature Scaling

````python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
````

# Training the training dataset

````python
from sklearn.naive_bayes import GaussianNB
GNB = GaussianNB()
GNB.fit(X_train,y_train)
````

# Predicting a new result

````python
print(GNB.predict(sc.transform([[30,87000]])))
````

# Gaussian Naive Bayes - Complete Working (Part 1: Training)

Gaussian Naive Bayes is a classification algorithm that learns the pattern of every class separately. During training, it does not memorize every customer. Instead, it learns a few statistical values from each class so that it can later calculate the probability of a new customer belonging to every class.

---

# Dataset

We are using the **Social_Network_Ads.csv** dataset.

| Age | EstimatedSalary | Purchased |
|----:|----------------:|----------:|
|19|19000|0|
|35|20000|0|
|26|43000|0|
|27|57000|0|
|...|...|...|
|42|149000|1|
|46|22000|1|
|47|49000|1|
|48|41000|1|

Features

- Age
- EstimatedSalary

Target

- Purchased

```
0 = Not Purchased
1 = Purchased
```

The goal is to learn the relationship between **Age**, **Estimated Salary**, and **Purchased**.

---

# Step 1: Import the Model

```python
from sklearn.naive_bayes import GaussianNB
```

Nothing is learned here.

The model does not know anything about the dataset yet.

# Step 5: Calculate Prior Probability

The model counts how many customers belong to each class.

Suppose

```

Class 1 = 120 customers

```

The Prior Probability is calculated as

```
P(Class 0)

180 / 300

```

```
P(Class 1)

=

120 / 300

=

0.40
```

The model stores these values.

These probabilities tell the model how common each class is before looking at any feature values.

# Step 6: Calculate Mean of Every Feature

Now the model starts learning the features.

First,

Example

```
19

35

26

27
```

Mean Age

```
=

(19+35+26+27)

÷4

26.75
```

The model stores

```
Mean Age (Class 0)

=

26.75
```

Next,

it repeats the same process for Salary.

```
19000

20000

43000

57000
```

It calculates

```
Mean Salary (Class 0)
```

Now it moves to Class 1.

Again,

it calculates

```
Mean Age (Class 1)

Mean Salary (Class 1)
```

After this step,

every feature of every class has its own Mean.

---

# Step 7: Calculate Variance of Every Feature

After calculating the Mean,

the model calculates the Variance.

For Age of Class 0,

it uses

```
Age Values

↓

19

35

26

27
```

and the Mean

```
26.75
```

It calculates how far every Age is from the Mean.

```
19 is far from 26.75

35 is also far

26 is very close

27 is very close
```

Using these distances,

the Variance is calculated.

The model stores

```

```

Then it repeats the same process for

```
Salary (Class 0)

↓

Age (Class 1)

↓

Salary (Class 1)
```

Now every feature has

```
Mean

Variance
```

stored separately for every class.

# Step 8: Why Mean and Variance Are Stored

Suppose a new customer arrives.

```
Age = 30

Salary = 87000

```

```

Variance
```

The Gaussian formula returns

```

```

Then the model again uses

```

Variance Salary
```

```
P(Salary | Class 0)
```

The same calculations are performed for Class 1.

So after training,

GaussianNB

The learning process is now complete.

The model is ready to predict new customers.

After training, the model has already learned

- Prior Probability of each class
- Mean of every feature for every class
- Variance of every feature for every class

Now a new customer arrives.

```python

```

# Step 9: Scale the New Customer

Before prediction, the customer must be scaled because the model was trained using scaled data.

# Step 10: Calculate Probability of Age for Class 0

The model first considers **Class 0 (Not Purchased).**

From training, it already knows

```
Mean Age (Class 0)

Variance Age (Class 0)
```

Now it substitutes

```
Age = -0.41

Mean Age (Class 0)

Variance Age (Class 0)
```

into the Gaussian Probability Density Formula.

The Gaussian formula returns

```
P(Age | Class 0)
```

This probability answers

> If the customer belongs to Class 0, how likely is it to observe this Age?

Suppose the returned value is

```
P(Age | Class 0)

=

0.31
```

The model stores this probability.

---

# Step 11: Calculate Probability of Salary for Class 0

Next,

the model repeats exactly the same process for Salary.

It already knows

```
Mean Salary (Class 0)

Variance Salary (Class 0)
```

It substitutes

```

Variance Salary (Class 0)
```

into the Gaussian formula.

```

```

=

0.31
```

```

=

0.18
```

---

# Step 12: Combine Feature Probabilities

Gaussian Naive Bayes assumes that the features are independent.

Therefore,

```
P(Features | Class 0)

```

```
0.31

×

0.18

```

```

=

0.0558
```

This value is called the **Likelihood**.

```

=

# Bayes' Theorem - Complete Explanation

Bayes' Theorem is a mathematical formula used to calculate the probability of an event after observing some new evidence.

It helps us answer questions like:

> "Now that I have some evidence, what is the probability that a particular event has happened?"

In Machine Learning, Bayes' Theorem is mainly used for **classification problems**, where the model predicts which class a new data point belongs to.

---

# Bayes' Theorem Formula

```
                 P(B|A) × P(A)
P(A|B) = ------------------------------
                  P(B)
```

Before understanding the formula, let us understand every probability separately.

---

# Step 1: What is an Event?

An **event** is simply something that may happen.

Examples

- It rains today.
- A student passes an exam.
- A customer purchases a product.
- A patient has cancer.

We usually represent events using capital letters.

```
A

B

C
```

---

# Step 2: What is P(A)?

```
P(A)
```

P(A) means

> **Probability of Event A occurring.**

It tells us how likely Event A is **before looking at any new evidence**.

Therefore,

P(A) is called the **Prior Probability**.

---

## Example

Suppose a hospital has

```
1000 patients
```

Out of them,

```
50 patients have cancer.
```

Then

```
P(Cancer)

=

50 / 1000

=

0.05
```

This means

There is a **5% chance** that a randomly selected patient has cancer **before looking at any medical test**.

---

# Step 3: What is P(B)?

```
P(B)
```

P(B) means

> **Probability of Event B occurring.**

In Bayes' Theorem,

Event B is usually the **new evidence** that we observe.

---

## Example

Suppose

```
B = Test Result is Positive
```

Then

```
P(Positive Test)
```

means

> What is the probability that a randomly selected patient gets a positive test result?

Notice that this includes

- Patients who actually have cancer.
- Patients who do not have cancer but still receive a positive test.

Therefore,

P(B) represents the probability of observing the evidence.

It is also called the **Evidence**.

---

# Step 4: What is P(A|B)?

```
P(A|B)
```

The symbol

```
|
```

is read as

```
given
```

Therefore,

```
P(A|B)
```

means

> **Probability of Event A given that Event B has already happened.**

This probability is called the **Posterior Probability** because it is calculated **after observing the evidence**.

---

## Example

Suppose

```
A = Patient has Cancer

B = Test Result is Positive
```

Then

```
P(Cancer | Positive Test)
```

means

> What is the probability that the patient has cancer **given that the test result is already positive?**

This is usually the probability we want to calculate.

---

# Step 5: What is P(B|A)?

```
P(B|A)
```

means

> **Probability of Event B given that Event A has already happened.**

This probability is called the **Likelihood**.

---

## Example

```
P(Positive Test | Cancer)
```

means

> If the patient already has cancer, what is the probability that the test becomes positive?

Suppose

```
100 cancer patients
```

Out of them

```
95 patients receive a positive test.
```

Therefore,

```
P(Positive Test | Cancer)

=

95 / 100

=

0.95
```

This means

If a patient has cancer,

there is a **95% chance** that the medical test becomes positive.

---

# Difference Between P(A|B) and P(B|A)

These two probabilities are **NOT the same.**

| Probability | Meaning |
|-------------|---------|
| P(A\|B) | Probability of A after knowing B |
| P(B\|A) | Probability of B after knowing A |

---

## Example

```
P(Cancer | Positive Test)
```

asks

> The test is already positive.

Now,

what is the probability that the patient has cancer?

---

```
P(Positive Test | Cancer)
```

asks

> The patient already has cancer.

Now,

what is the probability that the medical test becomes positive?

These are two completely different questions.

---

# How Bayes' Theorem Works

Bayes' Theorem combines all four probabilities.

```
                 P(B|A) × P(A)
P(A|B) = ------------------------------
                  P(B)
```

Now let us understand the role of every term.

---

## P(A)

This is the **Prior Probability**.

It represents our initial belief before seeing any evidence.

---

## P(B|A)

This is the **Likelihood**.

It tells us how likely the observed evidence is if Event A is actually true.

---

## P(B)

This is the **Evidence**.

It represents how likely the observed evidence is overall.

---

## P(A|B)

This is the **Posterior Probability**.

It is the final probability after combining

- Prior Knowledge
- New Evidence

This is the value Bayes' Theorem calculates.

---

# How Bayes' Theorem Works in Machine Learning

Suppose we are using the Social Network Ads dataset.

We define

```
A = Customer Purchased the Product
```

and

```
B = Customer's Features

Age

Estimated Salary
```

Now Bayes' Theorem becomes

```
                         P(Features | Purchased) × P(Purchased)
P(Purchased | Features) = -------------------------------------
                               P(Features)
```

Now understand every term.

---

# P(Purchased)

This is the Prior Probability.

It answers

> Before looking at the customer's Age and Salary,

what is the probability that a customer purchases the product?

The model calculates this during training by counting.

Example

```
Purchased = 1

120 customers

Total customers

320
```

Therefore

```
P(Purchased)

=

120 / 320
```

---

# P(Features)

The features are

```
Age

Estimated Salary
```

Suppose the new customer is

```
Age = 30

Estimated Salary = 87000
```

Then

```
P(Features)
```

means

> What is the probability of observing a customer having these feature values?

This is called the **Evidence**.

In Gaussian Naive Bayes,

this value is the same for every class comparison,

so it is usually ignored while comparing classes.

---

# P(Features | Purchased)

This is the **Likelihood**.

It answers

> If the customer has already purchased the product,

how likely is it to observe these feature values?

Since there are two features,

Gaussian Naive Bayes calculates

```
P(Age | Purchased)

×

P(Estimated Salary | Purchased)
```

using the stored Mean and Variance of the Purchased class.

---

# P(Purchased | Features)

This is the **Posterior Probability**.

It answers

> Given the customer's Age and Estimated Salary,

what is the probability that the customer will purchase the product?

This is the probability the model wants to calculate.

---

# How Gaussian Naive Bayes Uses Bayes' Theorem

During training (`fit()`), the model learns

- Prior Probability of every class.
- Mean of every feature for every class.
- Variance of every feature for every class.

These values are stored.

---

During prediction,

suppose a new customer arrives.

```
Age = 30

Estimated Salary = 87000
```

The model performs these steps.

### Step 1

Calculate

```
P(Age | Class 0)

P(Salary | Class 0)
```

using the Mean and Variance of Class 0.

---

### Step 2

Calculate

```
P(Age | Class 1)

P(Salary | Class 1)
```

using the Mean and Variance of Class 1.

---

### Step 3

Calculate

```
Posterior(Class 0)

=

Likelihood(Class 0)

×

Prior(Class 0)
```

---

### Step 4

Calculate

```
Posterior(Class 1)

=

Likelihood(Class 1)

×

Prior(Class 1)
```

---

### Step 5

Compare both Posterior Probabilities.

Example

```
Posterior(Class 0)

=

0.0048
```

```
Posterior(Class 1)

=

0.0019
```

Since

```
0.0048

>

0.0019
```

the model predicts

```
Class 0

↓

Not Purchased
```

---

# Summary

| Probability | Name | Meaning |
|-------------|------|---------|
| **P(A)** | Prior Probability | Probability of Event A before observing any evidence. |
| **P(B)** | Evidence | Probability of observing Event B. |
| **P(B\|A)** | Likelihood | Probability of observing Event B when Event A has already happened. |
| **P(A\|B)** | Posterior Probability | Probability of Event A after observing Event B. This is the final probability calculated using Bayes' Theorem. |

---

# Complete Working Flow

```
Choose the Event to Predict

↓

Find Prior Probability

P(A)

↓

Observe New Evidence

P(B)

↓

Calculate Likelihood

P(B|A)

↓

Apply Bayes' Theorem

↓

Calculate Posterior Probability

P(A|B)

↓

Compare Posterior Probabilities of All Classes

↓

Choose the Class with the Highest Probability

↓

Return Final Prediction
```
