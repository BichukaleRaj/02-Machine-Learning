## Data Preprocessing

- Data preprocessing is a collection of techniques used to prepare data for Machine Learning.
- A dataset **does not need every preprocessing technique**.
- Apply only the techniques required by your dataset.
- Example: If there are missing values → Fill them. If there is categorical data → Encode it.
- After preprocessing, split the data into Training and Testing sets.
- Then apply Feature Scaling (if required).

# Import libraries

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

````python
dataset = pd.read_csv('Data.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
````

# Taking care of missing data

````python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, 1:3])
X[:, 1:3] = imputer.transform(X[:, 1:3])
````

````python
print(X)
````

# Encoding categorical data

## Encoding the independent variable

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
````

````python
print(X)
````

## Encoding the dependent variable

````python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y =le.fit_transform(y)
````

````python
print(y)
````

# Splitting the dataset into the training set and test set

 ## Need of Splitting Data into Training Set and Test Set

### Why Split the Dataset?

If we train and test the model on the same data, the model may simply memorize the answers.

This gives unrealistically high accuracy.

---

### Training Set

Used to **train** the machine learning model.

```text
Model learns patterns from this data.
```

---

### Test Set

Used to **test** the trained model on unseen data.

```text
Checks how well the model performs on new data.
```

---

### Example

Suppose we have 100 records.

```text
80 Records → Training Set
20 Records → Test Set
```

The model learns from the 80 records and is tested on the remaining 20 records.

---

### Memory Trick 🧠

```text
Training Set = Learn
Test Set = Check
```

---

### Importance

- Prevents overfitting
- Evaluates real-world performance
- Tests model on unseen data
- Gives a fair measure of accuracy

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
````

# Feature Scaling

## Standardization

$$
x_{stand}=\frac{x-\operatorname{mean}(x)}{\operatorname{standard\ deviation}(x)}
$$

---

## Normalization

$$
x_{norm}=\frac{x-\min(x)}{\max(x)-\min(x)}
$$

````python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train[:, 3:] = sc.fit_transform(X_train[:, 3:])
X_test[:, 3:] = sc.transform(X_test[:, 3:])
````

## StandardScaler

```python
sc = StandardScaler()
```

- Standardizes numerical features.
- Makes feature values have:
  - Mean ≈ 0
  - Standard Deviation ≈ 1

```python
X_train[:,3:] = sc.fit_transform(X_train[:,3:])
X_test[:,3:] = sc.transform(X_test[:,3:])
```

- `fit_transform()` → Learn + Scale (Training Data)
- `transform()` → Scale using the same values (Test Data)

Only numerical columns are scaled.

## Why is my manual calculation different from `StandardScaler`?

The **formula is always the same**:

$$
x_{stand}=\frac{x-\text{mean}}{\text{standard deviation}}
$$

The difference is **where the Mean and Standard Deviation come from.**

### Manual Calculation

```text
Mean & Standard Deviation → Entire Dataset
```

### StandardScaler in Machine Learning

```text
Mean & Standard Deviation → Training Data (X_train) only
```

Then the same values are used to scale both:

```text
X_train
X_test
```

### Memory Trick 🧠

```text
Formula never changes.

Only the Mean and Standard Deviation change.
```

## `fit()`

```python
sc.fit(X_train)
```

👉 **Observes the training data and learns its Mean & Standard Deviation.**

```text
Learns only.
Does not change the data.
```

---

## `fit_transform()`

```python
X_train = sc.fit_transform(X_train)
```

👉 **Learns the Mean & Standard Deviation from X_train and immediately scales X_train.**

```text
Learn + Scale
```

---

## `transform()`

```python
X_test = sc.transform(X_test)
```

👉 **Uses the Mean & Standard Deviation learned from X_train to scale X_test.**

```text
Does not learn anything new.
Only scales.
```

---

### Memory Trick 🧠

```text
fit()            → Learn

fit_transform()  → Learn + Scale

transform()      → Scale using what was learned
```

# Why do we use the Mean & Standard Deviation of the **Training Data** for the **Test Data**?

After splitting the dataset:

- **Training Data** is used to **learn**.
- **Testing Data** is used to **evaluate** the trained model.

`StandardScaler` learns only from the **training data** by calculating:

- Mean
- Standard Deviation

These values are then used to standardize **both** the training and testing data.

If we calculate a new Mean and Standard Deviation from the **testing data**, then the preprocessing step has **learned from the test data**. As a result, the test data is no longer completely unseen, making the evaluation unfair.

Therefore:

```text
fit(X_train)
        ↓
Learn Mean & Standard Deviation

transform(X_train)
transform(X_test)
        ↓
Use the SAME Mean & Standard Deviation
```

### Memory Trick 🧠

```text
Learn once from X_train.

Use the same learning for:
✔ X_train
✔ X_test

Never learn from X_test.
```
