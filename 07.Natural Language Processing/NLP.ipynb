# Bag-of-Words

# Understanding Bag of Words using Set Theory

Think of the **Vocabulary** as the **Universal Set (Sample Space)**.

Example Vocabulary:

```text
V = {Apple, Mango, Banana, Orange, Grapes}
```

This is called the **Vocabulary** because it contains all the unique words found in the training dataset.

---

## Consider a Sentence

```text
Apple Mango Apple
```

The unique words used in this sentence are:

```text
S = {Apple, Mango}
```

At first, it may seem that Bag of Words simply creates a **subset** of the vocabulary.

But that is **not completely correct**.

---

## Why is it NOT a Subset?

A **Set** removes duplicate elements.

Example:

```text
Apple Mango Apple
```

As a Set:

```text
{Apple, Mango}
```

Notice that the second **Apple** disappears.

However, Bag of Words must remember that **Apple appeared twice**.

---

## What Bag of Words Actually Does

Instead of creating a subset, Bag of Words counts how many times each vocabulary word appears.

| Vocabulary Word | Count |
|-----------------|------:|
| Apple           | 2 |
| Mango           | 1 |
| Banana          | 0 |
| Orange          | 0 |
| Grapes          | 0 |

The Bag of Words vector becomes:

```text
[2, 1, 0, 0, 0]
```

---

## Mathematical Interpretation

Bag of Words can be thought of as a **frequency table**.

For every word in the vocabulary, we store:

```text
Frequency(word) = Number of times that word appears in the sentence
```

Example:

```text
Frequency(Apple)   = 2
Frequency(Mango)   = 1
Frequency(Banana)  = 0
Frequency(Orange)  = 0
Frequency(Grapes)  = 0
```

These frequencies are stored as:

```text
[2, 1, 0, 0, 0]
```

---

## Better Intuition

Instead of thinking of Bag of Words as a **subset**, think of it as a **frequency vector (histogram)**.

Flow:

```text
Vocabulary (Universal Set)
        ↓
Sentence uses some words
        ↓
Count the occurrence of EVERY vocabulary word
        ↓
Store those counts as a vector
```

---

## Why is it Called "Bag" of Words?

A **Bag (Multiset)** is similar to a Set, but **duplicates are allowed**.

Example:

Set

```text
{Apple, Mango}
```

Bag (Multiset)

```text
{Apple, Apple, Mango}
```

A Bag remembers **how many times** each element appears, while a Set only remembers **whether it exists or not**.

That is why the name **Bag of Words** is mathematically correct.

---

## Easy Summary

- Vocabulary is like the **Universal Set (Sample Space)**.
- A sentence contains some words from the vocabulary.
- Bag of Words **does not create a subset**, because subsets remove duplicates.
- Instead, it counts the frequency of every vocabulary word.
- These frequencies form a numerical vector.
- This vector is then given to Machine Learning algorithms like Logistic Regression, Naive Bayes, SVM, or Neural Networks.

# Import Libraries

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

# Import Dataset

````python
dataset = pd.read_csv("Restaurant_Reviews.tsv",delimiter ='\t',quoting=3)
````

````python
dataset
````

# Cleaning the text

````python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
corpus = []
for i in range(0, 1000):
  review = re.sub('[^a-zA-Z]', ' ', dataset['Review'][i])
  review = review.lower()
  review = review.split()
  ps = PorterStemmer()
  all_stopwords = stopwords.words('english')
  all_stopwords.remove('not')
  review = [ps.stem(word) for word in review if not word in set(all_stopwords)]
  review = ' '.join(review)
  corpus.append(review)
````

````python
corpus
````

# Creating Bag of Words

````python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
X = cv.fit_transform(corpus).toarray()
y = dataset.iloc[:,-1].values
````

````python
len(X[0])
````

# Splitting the data into Training set and Testing set 

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=0)
````

# Training the the Naive Bayes model on the Training set

````python
from sklearn.naive_bayes import GaussianNB
GNB = GaussianNB()
GNB.fit(X_train,y_train)
````

# Predicting the Test set results

````python
y_pred=GNB.predict(X_test)
````

# Making the confusion matrix

````python
from sklearn.metrics import confusion_matrix,accuracy_score
cm=confusion_matrix(y_test,y_pred)
ascore=accuracy_score(y_test,y_pred)
````
