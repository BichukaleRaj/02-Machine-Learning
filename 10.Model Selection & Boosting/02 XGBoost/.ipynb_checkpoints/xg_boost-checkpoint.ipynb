# XGBoost

## Importing the libraries

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

## Importing the dataset

````python
dataset = pd.read_csv('Data.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
````

## Splitting the dataset into the Training set and Test set

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
````

## Training XGBoost on the Training set

````python
from xgboost import XGBClassifier
classifier = XGBClassifier()
classifier.fit(X_train, y_train)
````

## Making the Confusion Matrix

````python
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
````

## Applying k-Fold Cross Validation

````python
from sklearn.model_selection import cross_val_score
accuracies = cross_val_score(estimator = classifier, X = X_train, y = y_train, cv = 10)
print("Accuracy: {:.2f} %".format(accuracies.mean()*100))
print("Standard Deviation: {:.2f} %".format(accuracies.std()*100))
````

# Training XGBoost on the Training Set (Using the Breast Cancer Dataset)

## Why do we use XGBoost?

In this problem, our goal is to predict whether a patient's tumor is **Benign** or **Malignant**.

The dataset contains information about the cells collected from a patient's breast tissue.

Each row represents **one patient**.

Each column represents one characteristic of the cells.

| Clump Thickness | Cell Size | Cell Shape | Bare Nuclei | ... | Class |
|-----------------|-----------|------------|--------------|-----|-------|
| 5 | 1 | 1 | 1 | ... | 2 |
| 5 | 4 | 4 | 5 | ... | 4 |
| 3 | 1 | 1 | 1 | ... | 2 |

Target Column

```
Class

2 = Benign (No Cancer)

4 = Malignant (Cancer)
```

Instead of manually finding patterns in these medical features, we train an XGBoost model so that it can automatically learn these patterns and predict the class of new patients.

---

# Step 1 : Import XGBoost

```python
from xgboost import XGBClassifier
```

This line imports the **XGBoost Classification** algorithm from the XGBoost library.

At this stage,

- no dataset is loaded into the model,
- no learning happens,
- no prediction is made.

Python only loads the XGBoost algorithm into memory so that we can use it.

---

# Step 2 : Create the XGBoost Model

```python
classifier = XGBClassifier()
```

Here Python creates an **empty XGBoost model**.

Think of it like creating a new student.

The student knows **how to learn**, but has **not studied anything yet**.

Currently,

- no Decision Trees exist,
- no rules have been learned,
- no predictions can be made.

The model is completely untrained.

---

# Step 3 : Start Training

```python
classifier.fit(X_train, y_train)
```

This is where the actual learning begins.

Earlier, we divided our dataset into

```
Training Set

and

Testing Set
```

Suppose after Train-Test Split,

```
Training Set = 512 patients

Testing Set = 171 patients
```

Python now gives the **512 training patients** to the XGBoost model.

Each patient looks something like this.

| Clump Thickness | Cell Size | Cell Shape | Bare Nuclei | ... | Class |
|-----------------|-----------|------------|--------------|-----|-------|
| 5 | 1 | 1 | 1 | ... | 2 |
| 8 | 10 | 10 | 8 | ... | 4 |
| 3 | 2 | 1 | 2 | ... | 2 |

Now XGBoost starts learning from these patients.

---

# Step 4 : Build the First Decision Tree

The first thing XGBoost does is create **one Decision Tree**.

The tree looks at all the medical features such as

- Clump Thickness
- Cell Size
- Cell Shape
- Bare Nuclei
- Bland Chromatin
- Mitoses

and tries to find patterns that separate

```
Benign

and

Malignant
```

patients.

For example,

the tree may learn rules like

```
If Cell Size is very large

↓

Possibility of Malignant increases.
```

or

```
If Bare Nuclei value is very small

↓

Possibility of Benign increases.
```

The first tree is built using all the training patients.

---

# Step 5 : Make Predictions Using the First Tree

After creating the first Decision Tree,

XGBoost asks the tree to predict the class of every training patient.

Suppose

Patient 1

```
Actual Class

Benign (2)
```

Prediction

```
Benign (2)
```

Correct.

---

Patient 2

```
Actual Class

Malignant (4)
```

Prediction

```
Malignant (4)
```

Correct.

---

Patient 3

```
Actual Class

Malignant (4)
```

Prediction

```
Benign (2)
```

Wrong.

---

Patient 4

```
Actual Class

Benign (2)
```

Prediction

```
Malignant (4)
```

Wrong.

Suppose

```
512 patients
```

were used for training.

The first tree predicts

```
470 patients correctly

42 patients incorrectly.
```

Most Machine Learning algorithms would stop here.

But XGBoost does **not** stop.

---

# Step 6 : Find the Mistakes

Now XGBoost carefully checks

```
Which patients were predicted incorrectly?
```

For example,

Patient

```
Actual

Malignant
```

Prediction

```
Benign
```

This patient becomes important.

Another patient

```
Actual

Benign
```

Prediction

```
Malignant
```

Again,

XGBoost remembers this mistake.

Instead of ignoring these mistakes,

XGBoost focuses on them.

This is the main idea behind Boosting.

---

# Step 7 : Build the Second Decision Tree

Now XGBoost creates another Decision Tree.

This second tree is **not built randomly**.

Its main purpose is

```
Correct the mistakes made by the first tree.
```

It pays much more attention to the patients that were predicted incorrectly.

Because of this,

the second tree learns patterns that the first tree failed to learn.

---

# Step 8 : Make Predictions Again

Now both trees work together.

The first tree gives its prediction.

The second tree tries to correct the mistakes of the first tree.

Suppose many of the previously wrong patients are now classified correctly.

The overall prediction becomes better.

---

# Step 9 : Build the Third Decision Tree

XGBoost again checks

```
Are there still any wrong predictions?
```

If the answer is yes,

another Decision Tree is created.

Again,

this tree mainly focuses on the remaining mistakes.

---

# Step 10 : Continue Building More Trees

The same process continues.

```
Build Tree 1

↓

Predict

↓

Find Mistakes

↓

Build Tree 2

↓

Predict Again

↓

Find Remaining Mistakes

↓

Build Tree 3

↓

Predict Again

↓

Repeat...
```

Instead of using only one Decision Tree,

XGBoost creates many trees.

Every new tree becomes better because it learns from the mistakes made by the previous trees.

---

# Step 11 : Combine All Decision Trees

After all the trees have been created,

XGBoost combines their predictions.

Suppose

Tree 1 predicts

```
Malignant
```

Tree 2 predicts

```
Malignant
```

Tree 3 predicts

```
Benign
```

Tree 4 predicts

```
Malignant
```

Instead of trusting only one tree,

XGBoost combines the predictions of all trees to produce one final prediction.

This combined prediction is usually much more accurate than using a single Decision Tree.

---

# Step 12 : Training is Complete

After learning from all training patients,

the XGBoost model is now trained.

It has learned many useful medical patterns such as

- Large Cell Size may indicate Malignant tumors.
- Very small Bare Nuclei values often indicate Benign tumors.
- Different combinations of features help distinguish between the two classes.

These learned patterns are stored inside the model.

Now the model is ready to predict completely new patients.

---

# Step 13 : Predict the Test Set

```python
y_pred = classifier.predict(X_test)
```

Now Python gives the **171 unseen patients** from the Test Set to the trained XGBoost model.

These patients were **never used during training**.

The model studies their medical features and predicts

```
2 = Benign

or

4 = Malignant
```

for every patient.

These predictions are stored inside

```python
y_pred
```

---

# Step 14 : Create the Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
```

Now Python compares

```
Actual Class

with

Predicted Class
```

for every patient.

Using this comparison,

it creates the **Confusion Matrix**, which shows

- Correct Benign Predictions
- Correct Malignant Predictions
- Wrong Benign Predictions
- Wrong Malignant Predictions

This helps us understand exactly where the model made mistakes.

---

# Step 15 : Calculate Accuracy

```python
accuracy_score(y_test, y_pred)
```

Python calculates

```
Accuracy

=

Correct Predictions

/

Total Predictions
```

Suppose

```
167 patients

correctly predicted
```

and

```
171 total patients
```

Then

```
Accuracy

=

167 / 171

=

97.66%
```

This tells us how well the XGBoost model performed on unseen patients.

---

# Step 16 : Perform K-Fold Cross Validation

```python
accuracies = cross_val_score(
    estimator = classifier,
    X = X_train,
    y = y_train,
    cv = 10
)
```

Instead of evaluating the model only once,

Python now performs **10-Fold Cross Validation**.

The training dataset is divided into

```
10 folds.
```

For each fold,

- 9 folds are used for training.
- 1 fold is used for validation.

This process repeats 10 times so that every fold becomes the validation set once.

Python stores all 10 accuracies.

For example

```
98%

97%

99%

98%

97%

98%

99%

98%

97%

98%
```

Now Python calculates

```python
accuracies.mean()
```

which gives the **Average Accuracy**.

It also calculates

```python
accuracies.std()
```

which tells us how much the accuracies changed across different folds.

A small Standard Deviation means the model is performing consistently.

---

# Key Points

- XGBoost is a **Boosting algorithm**.
- It creates **many Decision Trees**, not just one.
- The first tree learns the basic patterns.
- Every new tree focuses mainly on correcting the mistakes made by the previous trees.
- All trees work together to produce the final prediction.
- This makes XGBoost one of the most accurate Machine Learning algorithms for classification problems.
